## Part 1: PaddleOCR

PaddleOCR is a powerful, layout-aware OCR engine developed by Baidu. It's especially good at:
- Reading text in tables and complex layouts
- Detecting rotated or skewed text
- Providing precise polygon (not just rectangle) bounding boxes

Let's set it up!

```python
# ============================================
# STEP 1: Install PaddleOCR and All Dependencies
# ============================================
#
# WHAT WE'RE DOING:
# Installing PaddleOCR, PaddlePaddle, and helper libraries for PDF processing.
#
# ⚠️ IMPORTANT LOCAL EXECUTION NOTE:
# pdf2image requires the system library 'poppler' to be installed on your machine.
# - Ubuntu/Debian: sudo apt-get install poppler-utils
# - macOS: brew install poppler
# - Windows: Download poppler binaries and add them to your system PATH.
# ============================================

!pip install paddleocr==2.9.1 paddlepaddle==3.3.1 pdf2image langchain langchain-text-splitters

print("\n" + "=" * 50)
print("Python dependencies installed!")
print("Please ensure poppler is installed on your OS before proceeding.")
print("=" * 50)
```

```python
# ============================================
# STEP 2: Import Libraries
# ============================================

from paddleocr import PaddleOCR
from pdf2image import convert_from_path
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt
import cv2
import numpy as np
import os

print("All libraries imported successfully!")
```

```python
# ============================================
# STEP 3: Load Your Pharmaceutical SDF
# ============================================
# Ensure your PDF file is in the same directory as this notebook,
# or update the pdf_path variable below to point to your file.
# ============================================

pdf_path = "pharmaceutical-sdf-page4-certificate-processing.pdf"

if not os.path.exists(pdf_path):
    print(f"⚠️ Error: Could not find '{pdf_path}'. Please place the file in the correct directory.")
else:
    print(f"Processing: {pdf_path}\n")
    
    # Convert PDF to images (high quality)
    print("Converting PDF to image...")
    images = convert_from_path(pdf_path, dpi=300)
    print(f"Converted {len(images)} page(s)")
    
    # Save the first page as an image
    image = images[0]
    image_path = 'pharmaceutical_sdf.png'
    image.save(image_path, 'PNG')
    
    # Display the image
    plt.figure(figsize=(12, 8))
    plt.imshow(image)
    plt.title('Pharmaceutical SDF')
    plt.axis('off')
    plt.show()
```

```python
# ============================================
# STEP 4: Initialize PaddleOCR
# ============================================

print("Initializing PaddleOCR (this may download model files the first time)...")

ocr = PaddleOCR(
    use_angle_cls=True,
    lang='en'
)

print("\nPaddleOCR initialized and ready!")
```

```python
# ============================================
# STEP 5: Run PaddleOCR on the Pharmaceutical SDF
# ============================================

if 'image_path' in locals():
    print("Running PaddleOCR on your pharmaceutical document...")
    print("(This may take a moment)")
    
    result = ocr.ocr(image_path)
    
    print("\nOCR complete!")
    print(f"Detected {len(result[0]) if result[0] else 0} text regions")
else:
    print("Please run the PDF processing cell above first.")
```

```python
# ============================================
# STEP 6: Display Sample Results
# ============================================

if 'result' in locals() and result[0]:
    print("Sample of PaddleOCR Results:")
    print("=" * 60)
    
    # Show first 15 results
    for i, line in enumerate(result[0][:15]):
        box = line[0]  # Polygon coordinates
        text = line[1][0]  # Detected text
        confidence = line[1][1]  # Confidence score
        print(f"{i+1:2d}. [{confidence:.2%}] {text}")
        
    if len(result[0]) > 15:
        print(f"\n... and {len(result[0]) - 15} more text regions!")
elif 'result' in locals() and not result[0]:
    print("No text detected.")
else:
    print("Please run the OCR step above first.")
```
