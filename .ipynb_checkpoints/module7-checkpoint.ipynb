```python

# Create a list with categorical values.
l = ['B', 'A', 'B', 'C', 'C', 'A', 'A', 'B','A']

```

```python

# Display the contents of the list.
print(l)

```

```python

# Create a set out of l; only distinct values will be kept. 
s = set(l)

```

```python

# Display what element the set has.
print(s)

```

```python

# Create a list of keys. 
keys = list(s)

```

```python

print(keys)

```

```python

# Create a dictionary with keys and value 0s. 
dict = {}
for key in keys:
    dict[key] = 0

```

```python

# Each key in the dictionary has value 0. 
print(dict)

```

```python

print(dict.keys())

```

```python

# Traverse the list to find out the occurences of each key.
for key in dict.keys():
    for v in l:
        if key == v:
            dict[key] = dict[key] + 1

```

```python

# Display the dictionary. 
print(dict)

```

```python

# keys() returns the keys in the dictionary.
print(dict.keys())

```

```python

# values() returns the values in the dictionary.
print(dict.values())

```

```python

# Load the matplotlib package. 
import matplotlib.pyplot as plt

```

```python

# Draw a bar chart. 
X = ['A', 'B', 'C', 'D', 'F']
Y = [6, 12, 8, 4, 2]
plt.bar(X, Y)
plt.xlabel('Grades')
plt.ylabel('Counts')
plt.title("Grades Distribution")

```

```python

# Draw horizontal bars.
plt.barh(X, Y)
plt.xlabel('Counts')
plt.ylabel('Grades')
plt.title("Grades Distribution")

```

```python

# Draw a pie chart. 
plt.pie(Y, labels=X)

```

```python

# Display the percentage of each category. 
color_set = ['#ff0000', '#00ff00', '#0000ff', '#ffffAA', '#BC005A']
explode = (0,0.1,0, 0, 0)
plt.pie(Y, colors = color_set, labels = X,autopct='%1.1f%%', shadow = True, startangle= 90)

```

```python

# Emphasize a particular segment. 
color_set = ['#ff0000', '#00ff00', '#0000ff', '#ffffAA', '#BC005A']
explode = (0,0.1,0, 0, 0)
plt.pie(Y, colors = color_set, labels = X,autopct='%1.1f%%', shadow = True, startangle= 90, explode = explode)

```

```python

plt.pie(Y, labels=X)

```

```python

# Create a donut chart. 
plt.pie(Y, labels=X)
my_circle=plt.Circle( (0,0), 0.6, color='white')
p=plt.gcf()
p.gca().add_artist(my_circle)
plt.show()

```

```python

# load tips.csv into a dataframe. 
tips = pd.read_csv("tips.csv")

```

```python

len(tips)

```

```python

tips.head(10)

```

```python

# Group the data by day column, then by sex column; two-level grouping
tip_summary = tips.groupby(['day', 'sex'])['tip'].sum().unstack()

```

```python

print(tip_summary)

```

```python

# Draw a stacked bar chart. 
tip_summary.plot(kind='bar', stacked=True)
plt.title('Tips by Day and Gender')

```

```python

# Horizontal bar chart. 
tip_summary.plot(kind='barh', stacked=True)
plt.title('Tips by Day and Gender')

```

```python

# Bar chart to show sub groups. 
tip_summary.plot(kind='bar')

```

```python
# Draw a bar chart. 
X = video['Gender']
Y = video['Genre']
plt.bar(Y, X)
plt.xlabel('Grades')
plt.ylabel('Counts')
plt.title("Grades Distribution")
```

```python
video['Genre'].value_counts().plot(kind='bar')

# 2. Add labels and show the chart
plt.title('Distribution of Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```

```python
# Count values in the Genre column and plot a pie chart
video['Genre'].value_counts().plot(
    kind='pie',
    figsize=(8, 8),
    autopct='%1.1f%%',
    startangle=140,
    title='Genre Distribution',
)

plt.ylabel('')  # Removes the default 'Genre' label on the y-axis
plt.show()
```

```python
# 1. Group by Genre and Gender, then calculate the average Age
grouped_df = video.groupby(["Genre", "Gender"])["Age"].mean().unstack()

# 2. Plot the grouped bar chart
grouped_df.plot(kind="bar", figsize=(10, 6))

# 3. Customize the chart labels and title
plt.title("Average Age of Customers by Movie Genre and Gender")
plt.xlabel("Movie Genre")
plt.ylabel("Average Age")
plt.xticks(rotation=45)
plt.legend(title="Gender")
plt.tight_layout()

# 4. Display the chart
plt.show()
```
