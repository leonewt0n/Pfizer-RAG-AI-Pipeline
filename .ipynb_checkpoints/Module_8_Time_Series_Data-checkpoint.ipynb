```python
# Use date_range() in Pandas to create datetime objects. Per day by default. 
import pandas as pd
pd.date_range('8/1/2018', '8/31/2018')
```

```python
# Assign the dates to a variable.
Aug_2018 = pd.date_range('8/1/2018', '8/31/2018')
```

```python
Aug_2018
```

```python
type(Aug_2018)
```

```python
# Sea_levels contains 31 integers. 
sea_levels = [2	,
25	,
48	,
28	,
5	,
24	,
23	,
28	,
5	,
53	,
25	,
89	,
1	,
25	,
29	,
21	,
13	,
29	,
13	,
29	,
29	,
27	,
17	,
28	,
48	,
55	,
58	,
75	,
68	,
63	,
39	,
]
```

```python
# Create a dataframe with two columns; one contains dates and one contains integers.
from pandas import DataFrame
df = DataFrame({'Dates': Aug_2018, 'Level': sea_levels})
df.reset_index(drop=True, inplace=True)
```

```python
df
```

```python
# Draw a line chart out of the dataframe. 
df.plot.line(x='Dates', y = 'Level')
```

```python
# Computing the rolling window means. Forget to set the window size. 
df['Moving Avg'] = df['Level'].rolling().mean()
```

```python
# Window size is set to 3. Save the results to another column in dataframe. 
df['Moving Avg'] = df['Level'].rolling(window=3).mean()
```

```python
df
```

```python
# Plot the moving averages. 
df.plot.line(x='Dates', y = 'Moving Avg')
```

```python
# Compare the original time series data with the moving averages. 
# Notice the  orange line misses a small segment. Why? 
df.plot.line(x='Dates', y = ['Level', 'Moving Avg'])
```

```python
# Computing the rolling max value of 3 consecutive level values. 
df['Moving High'] = df['Level'].rolling(window=3).max()
```

```python
df
```

```python
# Show the original data with the rolling max values. 
df.plot.line(x='Dates', y = ['Level', 'Moving High'])
```

```python
# Compute the min value of every 3 consecutive levels.
df['Moving Low'] = df['Level'].rolling(window=3).min()
```

```python
df
```

```python
# Visualize the original data and the rolling low values. 
df.plot.line(x='Dates', y = ['Level', 'Moving Low'])
```

```python
# Compute the std of 3 consecutive level values. 
df['Moving Std'] = df['Level'].rolling(window=3).std()
```

```python
df
```

```python
# Visualize the original data and rolling std values. 
df.plot.line(x='Dates', y = ['Level', 'Moving Std'])
```

```python
# Load a csv file to a dataframe
import pandas as pd
close_px_all = pd.read_csv("stock_px.csv", parse_dates=True, index_col=0)
```

```python
# Use only the first 3 columns.
close_px = close_px_all[['AAPL', 'MSFT', 'XOM']]
```

```python
close_px.head(10)
```

```python
# Visualize all three stock prices.
close_px.plot()
```

```python
# Show only AAPL stock prices.
close_px.AAPL.plot()
```

```python
# Show AAPL stock prices with rolling mean, min, max, std values.
# The window size is 250; roughly about the number of trading days in a year. 
close_px['AAPL'].plot()
close_px.AAPL.rolling(window=250).mean().plot()
close_px.AAPL.rolling(window=250).min().plot()
close_px.AAPL.rolling(window=250).max().plot()
close_px.AAPL.rolling(window=250).std().plot()
```

```python
#Find the year with the largest total volume
amazon = pd.read_csv("amazon_10_years.csv",parse_dates=['Date'])
print(amazon.groupby(amazon['Date'].dt.year)['Volume'].sum().idxmax())
```

```python
#Weekday with the highest average volume number 
print(amazon.groupby(amazon['Date'].dt.day_name())['Volume'].mean().idxmax())
```
